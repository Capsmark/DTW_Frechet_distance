```python
import pandas as pd

# Define file paths
file_paths = [
    'E:\sigai\seperate-data\\1.csv',
    'E:\sigai\seperate-data\\2.csv',
    'E:\sigai\seperate-data\\3.csv',
    'E:\sigai\seperate-data\\4.csv',
    'E:\sigai\seperate-data\\5.csv',
    'E:\sigai\seperate-data\\6.csv',
    'E:\sigai\seperate-data\\7.csv'
]
# Function to standardize column names
def standardize_columns(df):
    if 'Sigai' in df.columns:
        df = df.rename(columns={'Sigai': 'sigai'})
    return df

# Read all datasets into a list of DataFrames
dfs = [standardize_columns(pd.read_csv(file)) for file in file_paths]

# Concatenate all DataFrames into a single DataFrame
combined_df = pd.concat(dfs)

# Ensure the datetime column is in datetime format
combined_df['datetime'] = pd.to_datetime(combined_df['datetime'])

# Remove duplicate datetime entries
combined_df = combined_df.drop_duplicates(subset=['datetime'])

# Sort the DataFrame by datetime
combined_df = combined_df.sort_values(by='datetime')

# Save the resulting DataFrame to a CSV file
combined_df.to_csv('E:\sigai\seperate-data\merged_dataset.csv', index=False)

print("Datasets merged and saved to E:\sigai\seperate-data\merged_dataset.csv")
```
